```python
!pip install -q transformers torch gradio

from google.colab import drive
drive.mount('/content/drive')

from transformers import AutoTokenizer, AutoModelForMaskedLM, pipeline
import torch

MODEL_PATH = '/content/drive/MyDrive/bert_cybersecurity_model'

!ls -lh /content/drive/MyDrive/bert_cybersecurity_model

tokenizer = AutoTokenizer.from_pretrained(MODEL_PATH)
model = AutoModelForMaskedLM.from_pretrained(MODEL_PATH)

device = 0 if torch.cuda.is_available() else -1

fill_mask = pipeline(
    'fill-mask',
    model=model,
    tokenizer=tokenizer,
    device=device
)

text = 'a computer [MASK] connects multiple devices together.'
results = fill_mask(text, top_k=5)

for result in results:
    print(f"الكلمة: {result['token_str']} | الاحتمالية: {result['score']:.4f}")
```

```python
import gradio as gr


def predict_masked_text(text):
    mask_token = tokenizer.mask_token

    if not text or text.count(mask_token) != 1:
        return {
            f'يرجى إدخال نص يحتوي على {mask_token} مرة واحدة فقط': 1.0
        }

    results = fill_mask(text, top_k=5)

    return {
        result['token_str'].strip(): float(result['score'])
        for result in results
    }


demo = gr.Interface(
    fn=predict_masked_text,
    inputs=gr.Textbox(
        lines=3,
        placeholder='مثال: firewall protects the [MASK] from unauthorized access',
        label='أدخل جملة تحتوي على [MASK]'
    ),
    outputs=gr.Label(
        num_top_classes=5,
        label='الكلمات المتوقعة'
    ),
    title='Cybersecurity BERT Language Model Demo',
    description='نموذج BERT مدرب على نصوص الشبكات والأمن السيبراني.',
    examples=[
        ['a computer [MASK] connects multiple devices together.'],
        ['firewall monitors incoming and outgoing network [MASK].'],
        ['ip address is used to identify devices on a [MASK].']
    ]
)

demo.launch(share=True)
```
