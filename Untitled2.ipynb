```python
from google.colab import drive
import pandas as pd

# 1. ربط Google Drive ببيئة Colab
drive.mount('/content/drive')


file_path = '/content/drive/MyDrive/NLP_Project/scraped_data.csv'
import pandas as pd

# 3. قراءة البيانات
df = pd.read_csv(file_path)

# 4. طباعة ملخص شامل للبيانات
print("=== إجمالي عدد الأسطر والأعمدة ===")
print(f"الأسطر: {df.shape[0]}, الأعمدة: {df.shape[1]}")

print("\n=== أسماء الأعمدة ===")
print(df.columns.tolist())

print("\n=== فحص القيم المفقودة (Nulls) ===")
print(df.isnull().sum())

print("\n=== عينة من أول 3 أسطر ===")
display(df.head(3))



```

```python
import re
import string
import nltk
from nltk.corpus import stopwords

# تنزيل حزم NLTK المطلوبة
nltk.download('stopwords')
nltk.download('punkt')

# 1. إزالة الأسطر المكررة والأنماط الفارغة
print(f"عدد الأسطر قبل إزالة التكرار: {len(df)}")
df = df.drop_duplicates(subset=['Text']).dropna(subset=['Text'])
print(f"عدد الأسطر بعد إزالة التكرار: {len(df)}")

# 2. تجهيز قائمة كلمات التوقف الإنجليزية
stop_words = set(stopwords.words('english'))

# 3. دالة تنظيف النص والتطبيع (Normalization)
def clean_text(text):
    # تحويل الأحرف إلى صغيرة (Lowercasing)
    text = str(text).lower()
    # إزالة الروابط ووسوم HTML
    text = re.sub(r'http\S+|www\S+|<.*?>', '', text)
    # إزالة علامات الترقيم والأرقام
    text = text.translate(str.maketrans('', '', string.punctuation)) # delete punctuation and ()
    text = re.sub(r'\d+', '', text) # delete numbers in text
    # تقطيع النص وإزالة كلمات التوقف (Stop Words)
    words = text.split()
    cleaned_words = [word for word in words if word not in stop_words and len(word) > 1]
    # إعادة تركيب النص المنظف
    return ' '.join(cleaned_words)

# تطبيق دالة التنظيف وإنشاء عمود جديد
df['cleaned_text'] = df['Text'].apply(clean_text)

# 4. عرض عينة مقارنة قبل وبعد التنظيف
print("\n=== عينة مقارنة للنصوص قبل وبعد التنظيف ===")
for i, row in df.head(2).iterrows():
    print(f"النص الأصلي {i+1}:\n{row['Text']}\n")
    print(f"النص بعد التنظيف {i+1}:\n{row['cleaned_text']}\n")
    print("-" * 60)
```

```python
!pip install -q transformers datasets

from transformers import AutoTokenizer

# 1. اختيار واستدعاء الـ Tokenizer الخاص بنموذج bert-base-uncased
model_name = "bert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(model_name)

# 2. اختيار عينة وتطبيق الترميز (Encoding)
sample_text = df['cleaned_text'].iloc[0]
encoded_sample = tokenizer(
    sample_text,
    max_length=128,
    padding='max_length',
    truncation=True,
    return_tensors="pt"
)

# 3. عرض نتائج التقطيع والترميز وفك الترميز (Decoding)
print("=== تجربة التقطيع والترميز (Encoding/Decoding) ===")
print("\n1. الرموز الناتجة (Tokens):")
print(tokenizer.convert_ids_to_tokens(encoded_sample['input_ids'][0])[:15])

print("\n2. الأرقام الممثلة للرموز (Input IDs):")
print(encoded_sample['input_ids'][0][:15])

print("\n3. قناع الانتباه (Attention Mask):")
print(encoded_sample['attention_mask'][0][:15])

print("\n4. إعادة فك الترميز للنص الأصلي (Decoding):")
print(tokenizer.decode(encoded_sample['input_ids'][0], skip_special_tokens=True))
```

```python
''''train using full data''''

from datasets import Dataset
from transformers import AutoModelForMaskedLM, DataCollatorForLanguageModeling, TrainingArguments, Trainer

# 1. تحويل Pandas DataFrame إلى Hugging Face Dataset
hf_dataset = Dataset.from_pandas(df[['cleaned_text']])

# 2. تقسيم البيانات إلى تدريب (90%) واختبار (10%)
hf_dataset = hf_dataset.train_test_split(test_size=0.1)

# 3. تطبيق الـ Tokenizer على البيانات بالكامل
def tokenize_function(examples):
    return tokenizer(examples['cleaned_text'], truncation=True, max_length=128)

tokenized_datasets = hf_dataset.map(tokenize_function, batched=True, remove_columns=['cleaned_text'])

# 4. تحميل نموذج bert-base-uncased لإعادة الضبط
model = AutoModelForMaskedLM.from_pretrained("bert-base-uncased")

# 5. إعداد تجميع البيانات وإخفاء الكلمات بنسبة 15% (Masked Language Modeling)
data_collator = DataCollatorForLanguageModeling(tokenizer=tokenizer, mlm_probability=0.15)

# 6. تحديد خيارات وإعدادات التدريب
training_args = TrainingArguments(
    output_dir="./bert_finetuned_network",
    eval_strategy="epoch",  # تم التحديث هنا إلى eval_strategy
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    num_train_epochs=1,
    weight_decay=0.01,
    save_strategy="epoch",
    logging_steps=100,
    report_to="none"
)

# 7. إنشاء الـ Trainer وبدء عملية الـ Fine-Tuning
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_datasets['train'],
    eval_dataset=tokenized_datasets['test'],
    data_collator=data_collator,
)

print("=== بدء عملية تدريب الضبط الدقيق (Fine-Tuning) ===")
trainer.train()
```

```python
import torch
from datasets import Dataset
from transformers import AutoModelForMaskedLM, DataCollatorForLanguageModeling, TrainingArguments, Trainer

# 1. اقتطاع عينة عشوائية ممثلة بـ 10,000 سطر
df_sampled = df.sample(n=10000, random_state=42).reset_index(drop=True)

# 2. تحويل البيانات وتقسيمها (90% تدريب / 10% اختبار)[span_1](start_span)[span_1](end_span)
hf_dataset = Dataset.from_pandas(df_sampled[['cleaned_text']])
hf_dataset = hf_dataset.train_test_split(test_size=0.1)

# 3. تطبيق التقطيع والترميز (Tokenization)[span_2](start_span)[span_2](end_span)
def tokenize_function(examples):
    return tokenizer(examples['cleaned_text'], truncation=True, max_length=128)

tokenized_datasets = hf_dataset.map(tokenize_function, batched=True, remove_columns=['cleaned_text'])

# 4. تحميل النموذج bert-base-uncased[span_3](start_span)[span_3](end_span)
model = AutoModelForMaskedLM.from_pretrained("bert-base-uncased")

# 5. إعداد تجميع البيانات وإخفاء الكلمات بنسبة 15% (MLM)
data_collator = DataCollatorForLanguageModeling(tokenizer=tokenizer, mlm_probability=0.15)

# 6. إعدادات التدريب المحسنة للسرعة والأداء High-Speed Settings[span_4](start_span)[span_4](end_span)
training_args = TrainingArguments(
    output_dir="./bert_finetuned_fast",
    eval_strategy="epoch",
    learning_rate=2e-5,
    per_device_train_batch_size=32,   # مضاعفة حجم الدفعة
    num_train_epochs=3,               # تدريب لـ 3 دورات كاملة
    fp16=torch.cuda.is_available(),   # تفعيل الدقة النصفية لتسريع التدريب
    weight_decay=0.01,
    save_strategy="epoch",
    logging_steps=50,
    report_to="none"
)

# 7. بدء عملية الضبط الدقيق (Fine-Tuning)[span_5](start_span)[span_5](end_span)
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_datasets['train'],
    eval_dataset=tokenized_datasets['test'],
    data_collator=data_collator,
)

print("=== بدء التدريب المحسّن (3 Epochs) ===")
trainer.train()
```

```python
!pip install -q gradio

import gradio as gr
from transformers import pipeline

# 1. استدعاء النموذج المرفوع من Hugging Face
fill_mask = pipeline("fill-mask", model=model_id)

# 2. دالة التنبؤ للواجهة
def predict_masked_text(text):
    if "[MASK]" not in text:
        return {"خطأ: يرجى تضمين الكلمة [MASK] داخل النص": 0.0}
    results = fill_mask(text)
    return {res['token_str']: float(res['score']) for res in results}

# 3. تصميم واجهة المستخدم التفاعلية
demo = gr.Interface(
    fn=predict_masked_text,
    inputs=gr.Textbox(
        lines=3,
        placeholder="أدخل نصاً يحتوي على [MASK] مثل: firewall protects the [MASK] from unauthorized access",
        label="Cybersecurity Input Sentence"
    ),
    outputs=gr.Label(num_top_classes=5, label="Top 5 Predicted Words & Probabilities"),
    title=" Cybersecurity BERT Language Model Demo",
    description="اختبار نموذج BERT المخصص لشبكات الكمبيوتر والأمن الرقمي بعد الضبط الدقيق والرفع على Hugging Face.",
    examples=[
        ["a computer [MASK] connects multiple devices together."],
        ["firewall monitors incoming and outgoing network [MASK]."],
        ["ip address is used to identify devices on a [MASK]."]
    ]
)

# 4. تشغيل الواجهة وتوليد رابط مشاركة عام (Public Link)
demo.launch(share=True)
```
